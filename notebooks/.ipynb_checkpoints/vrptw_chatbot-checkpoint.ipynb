# Chatbot VRPTW
Este notebook permite consultar as rotas dos veículos a partir do arquivo `melhor_solucao.json` e integrar com o LLM para gerar instruções detalhadas.

```python
import json
import os
from llm_integration.llm_integration import gerar_instrucoes_motorista

# Caminho absoluto da raiz do projeto
root_path = os.path.abspath(os.path.join(os.getcwd(), ".."))

# Caminho do JSON
json_path = os.path.join(root_path, "melhor_solucao.json")

# Carregar JSON da melhor solução
with open(json_path, 'r', encoding='utf-8') as f:
    data = json.load(f)

best_solution = data['melhor_solucao']
distancia_total = data['distancia_total']
instrucoes = data.get('instrucoes_motoristas')

if not instrucoes:
    # se não tiver instrucoes no JSON, gera via LLM
    instrucoes = gerar_instrucoes_motorista(best_solution, distancia_total)
```

```python
# Função simples de chatbot
def consultar_rota(texto):
    texto = texto.lower()
    # identificar veículo
    for rinfo in best_solution['routes']:
        vid = rinfo['vehicle_id']
        if f'veículo {vid}' in texto or f'veiculo {vid}' in texto:
            return instrucoes.get(vid, f'Nenhuma instrução para veículo {vid}')
    return 'Não consegui identificar o veículo na pergunta.'
```

```python
# Exemplos de perguntas
print(consultar_rota('Qual a melhor rota para o veículo 0?'))
print(consultar_rota('Me diga a rota do veiculo 1.'))
print(consultar_rota('Rota do veículo 2'))
```
