## 📂 Carregar `melhor_solucao.json`

```python

# Carregar melhor solução do JSON
with open("../melhor_solucao.json", "r", encoding="utf-8") as f:
    data = json.load(f)

best_solution = data["melhor_solucao"]
distancia_total = data["distancia_total"]
```

## 🧠 Função de consulta (LLM + fallback)

```python
def ask_llm(query_text):
    """
    Interpreta a pergunta do usuário e retorna uma resposta usando a função gerar_instrucoes_motorista.
    """
    query_text = query_text.lower()

    # Caso 1: Pergunta sobre rota específica
    if "rota" in query_text and "veículo" in query_text:
        import re
        match = re.search(r"ve[ií]culo\s*(\d+)", query_text)
        if match:
            vid = int(match.group(1)) - 1
            instrucoes = gerar_instrucoes_motorista(best_solution, distancia_total)
            return instrucoes.get(vid, "Veículo não encontrado.")

    # Caso 2: Pergunta sobre resumo geral
    elif "melhor rota" in query_text or "resumo" in query_text:
        instrucoes = gerar_instrucoes_motorista(best_solution, distancia_total)
        resumo = "\n\n".join(instrucoes.values())
        return f"Resumo das rotas:\n{resumo}"

    # Caso 3: Qualquer outra pergunta — resposta genérica
    else:
        instrucoes = gerar_instrucoes_motorista(best_solution, distancia_total)
        return "\n".join([f"🚚 {v}: {txt[:200]}..." for v, txt in instrucoes.items()])

    system = "Você é um assistente logístico que responde perguntas sobre rotas otimizadas. Responda em português e seja objetivo."
    user_prompt = f"Contexto:\n{context_text}\n\nPergunta: {question}\n\nResponda de forma clara e direta, com bullets quando fizer sentido."

    def simulated_responder(q):
        ql = q.lower()
        m = re.search(r"ve[ií]culo\s*(\d+)", ql)
        if m:
            vid = m.group(1)
            if vid in veh_map:
                v = veh_map[vid]
                text = f"Veículo {vid}: carga {v['load']}/{v['capacity']}, paradas {v['n_stops']}, distância ≈ {v['distance']:.1f}.\nRota: " + ' -> '.join([str(p) for p in v['route']])
                if v['over_capacity']>0:
                    text += f"\n⚠️ Excede capacidade em {v['over_capacity']}."
                return text
            else:
                return f"Não encontrei o veículo {vid}."
        if 'melhor rota' in ql or 'melhor solução' in ql:
            best = min(summaries, key=lambda x: x['distance'])
            return f"A rota com menor distância é do veículo {best['vehicle_id']} (~{best['distance']:.1f})."
        if 'mais carregado' in ql or 'maior carga' in ql:
            best = max(summaries, key=lambda x: x['load'])
            return f"O veículo mais carregado é o {best['vehicle_id']} com {best['load']} unidades."
        return "Desculpe, não entendi exatamente. Pergunte: 'Qual a rota do veículo 1?' ou 'Qual veículo está sobrecarregado?'"

    if client is None:
        return simulated_responder(question)

    try:
        resp = client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[
                {"role": "system", "content": system},
                {"role": "user", "content": user_prompt}
            ],
            temperature=0.2,
            max_tokens=400
        )
        return resp.choices[0].message.content
    except Exception as e:
        return f"Erro na API OpenAI: {e}\nUsando resposta simulada:\n" + simulated_responder(question)
```

## 💬 Interface de chat (ipywidgets)

```python
input_box = widgets.Text(placeholder="Digite sua pergunta...")
output_box = widgets.Output()

def on_submit(change):
    with output_box:
        print(f"🧑‍💼 Você: {change['new']}")
        resposta = ask_llm(change['new'])
        print(f"🤖 Assistente: {resposta}\n")

input_box.observe(on_submit, names='value')

display(input_box, output_box)
```
