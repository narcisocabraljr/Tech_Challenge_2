# 🤖 Demo LLM Chat — Otimização de Rotas


Este notebook demonstra um **chatbot LLM integrado** ao projeto de **otimização de rotas**.
Ele permite conversar com o assistente sobre as rotas encontradas pelo algoritmo genético,
simulando uma camada LLM de linguagem natural.


```python

import json
from IPython.display import display, Markdown
from llm_integration import gerar_instrucoes_motorista
```

```python
json_path = "melhor_solucao.json"

with open(json_path, "r", encoding="utf-8") as f:
    data = json.load(f)

melhor_solucao = data["melhor_solucao"]
distancia_total = data["distancia_total"]
instrucoes_motoristas = data.get("instrucoes_motoristas", None)
```

```python
def consulta_rota(pergunta: str):
    """
    Recebe uma pergunta tipo 'Qual a melhor rota para o veículo 1?'
    Retorna instruções detalhadas ou resumo da rota.
    """
    # tenta extrair veículo da pergunta
    import re
    match = re.search(r've[ií]culo (\d+)', pergunta, re.IGNORECASE)
    if not match:
        return "Desculpe, não consegui identificar o veículo na pergunta."
    
    vid = int(match.group(1))
    
    # verifica se o veículo existe
    veiculos = [r["vehicle_id"] for r in melhor_solucao["routes"]]
    if vid not in veiculos:
        return f"Veículo {vid} não encontrado na solução."

    # obtém a rota
    route_info = next(r for r in melhor_solucao["routes"] if r["vehicle_id"] == vid)
    route = route_info["route"]
    total_load = route_info["total_load"]
    capacity = route_info["capacity"]

    # se já houver instruções geradas pelo LLM, usa
    if instrucoes_motoristas and str(vid) in instrucoes_motoristas:
        return instrucoes_motoristas[str(vid)]

    # caso contrário, gera um resumo simples
    texto = f"Veículo {vid} (carga {total_load}/{capacity}) deve visitar as cidades na seguinte ordem:\n"
    for idx, city in enumerate(route):
        if idx == 0:
            texto += f"  - Depósito {city}\n"
        else:
            texto += f"  - Cidade {city}\n"
    texto += f"Distância aproximada total: {distancia_total:.1f} unidades"
    
    return texto
```

```python
while True:
    pergunta = input("Você: ")
    if pergunta.lower() in ["sair", "exit"]:
        break
    resposta = consulta_rota(pergunta)
    print(f"Bot: {resposta}\n")
```

## 🧠 Função simulando a LLM

```python

def llm_responder(pergunta, dados):
    pergunta = pergunta.lower()

    match = re.search(r"ve[ií]culo\s*(\d+)", pergunta)
    if match:
        vid = match.group(1)
        if vid in dados["veiculos"]:
            v = dados["veiculos"][vid]
            resposta = f"🚛 **Veículo {vid}**\n"
            resposta += f"- Capacidade: {v['capacidade']}\n"
            resposta += f"- Carga Atual: {v['carga']}\n"
            resposta += f"- Rota planejada:\n"
            for p in v["rota"]:
                resposta += f"  • {p['id']} ({p['tipo']}) em {p['coordenadas']}\n"
            return resposta
        else:
            return f"Não encontrei informações para o veículo {vid}."

    if "melhor rota" in pergunta or "melhor solução" in pergunta:
        return "A melhor solução foi calculada considerando distância total, capacidade e janelas de tempo. Você pode pedir detalhes de cada veículo, como 'Qual a rota do veículo 0?'."

    return "Desculpe, ainda não entendi. Tente perguntar algo como 'Qual a rota do veículo 1?' ou 'Mostre a melhor solução'."
```

## 💬 Interface do chat

```python

def iniciar_chat(dados):
    display(Markdown("### 🤖 Chat LLM — Otimização de Rotas"))
    print("Digite 'sair' para encerrar.\n")

    while True:
        pergunta = input("Você: ")
        if pergunta.lower() in ["sair", "exit", "quit"]:
            print("\nEncerrando o chat. 👋")
            break
        resposta = llm_responder(pergunta, dados)
        display(Markdown(resposta))
```

## ▶️ Executar o chat

```python
iniciar_chat(melhor_solucao)
```
